```python
import tables
import numpy as np
from ctapipe_io_lst.constants import LST1_LOCATION as location
from datetime import datetime, timezone


a = tables.open_file("/data/cta/users-ifae/moralejo/CTA/summer_student_2026/datacheck/datacheck_dl1_LST-1.Run24704.h5")
```

```python
subrun = 8

# ALTITUDE AND AZIMUTH ANGLES OF THE TEL
tel_alt = np.degrees(a.root.dl1datacheck.cosmics.col('mean_alt_tel')[subrun])
tel_az  = np.degrees(a.root.dl1datacheck.cosmics.col('mean_az_tel')[subrun])

# TEL LOCATION
lat = location.lat.deg
lon = location.lon.deg
height = location.height.to_value('m')

# DRAGON TIMES (UTC for 50 events through the run, given in Unix time)
dragon_times = a.root.dl1datacheck.cosmics.col('dragon_time')[subrun]

ti_unix = np.min(dragon_times)
tmean_unix = np.mean(dragon_times)
tf_unix = np.max(dragon_times)

# Conversion to datetime UTC
ti = datetime.fromtimestamp(ti_unix, tz=timezone.utc)
tmean = datetime.fromtimestamp(tmean_unix, tz=timezone.utc)
tf = datetime.fromtimestamp(tf_unix, tz=timezone.utc)

print("="*60)
print(f"SUBRUN {subrun}")
print("="*60)

print("\nTime (UTC)")
print(f"Initial : {ti.isoformat()}")
print(f"Mean  : {tmean.isoformat()}")
print(f"Final  : {tf.isoformat()}")
print(f"Interval: {tf_unix - ti_unix:.2f} s")

print("\nTelescope")
print(f"Altitude = {tel_alt:.4f} deg")
print(f"Azimuth  = {tel_az:.4f} deg")

print("\nLST1 Observatory")
print(f"Latitude  = {lat:.6f} deg")
print(f"Longitude = {lon:.6f} deg")
print(f"Height   = {height:.1f} m")

```
